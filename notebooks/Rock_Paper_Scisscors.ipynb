Python Project:

Make a two-player Rock-Paper-Scissors game. (Hint: Ask for player plays
(using input), compare them, print out a message of congratulations to the
winner, and ask if the players want to start a new game)
Remember the rules:
 Rock beats scissors
 Scissors beats paper
 Paper beats rock

```python

# Online Python - IDE, Editor, Compiler, Interpreter

while True:
    start = input("Lets play Rock Paper Scissors. Are you ready to play? /n Yes or No: " )
    if start.lower() == 'no':
        print('Ok see you next time')
        break
    elif start.lower() == 'yes':
        print("You selected Yes. Time to begin the game!")
        
        players = input('do you have 2 players? ')
        if players.lower() == 'no':
            print('You need 2 players ')
            break
        else:
            player1 = input('Player 1: Choose Rock Paper Scissors ')
            while player1.lower() not in ['rock', 'paper', 'scissors']:
                player1 = input("Player 1: Please choose Rock, Paper, or Scissors: ")
            print('Player 1 choose ' + player1)
            
            player2 = input('Player 2: Choose Rock Paper Scissors: ')
            while player2.lower() not in ['rock', 'paper', 'scissors']:
                player2 = input('Player 2: You mus choose either Rock Paper or Scissors: ')
            print('Player 2 choose ' + player2)
            
            if player1.lower() == player2.lower():
                print("It's a tie!")
            elif player1.lower() == 'rock':
                if player2.lower() == 'scissors':
                    print("Player 1 wins!")
                else:
                    print('Player 2 wins!')
            elif player1.lower() == 'paper':
                if player2.lower() == 'rock':
                    print('Player 1 wins!')
                else:
                    print('Player 2 wins!')
            elif player1.lower() == 'scissors':
                if player2.lower() == 'paper':
                    print("Player 1 wins!")
                else:
                    print('Player 2 wins!')
        
            new_game = input('Play again? Yes or No ')
            if new_game.lower() != 'yes':
                print('Game over')
                break
                        
    else:
        print('You have not selected yes or no good bye')
        break

        
                        
            
                
```
